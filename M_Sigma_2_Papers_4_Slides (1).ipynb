# Die M–σ-Beziehung
## Schwarze-Loch-Masse und stellare Geschwindigkeitsdispersion

**Leitfrage:** Wie hängt die Masse des zentralen supermassereichen Schwarzen Lochs mit der Geschwindigkeitsdispersion der Sterne seiner Wirtsgalaxie zusammen?

Literaturreview zu zwei Arbeiten:

- **Graham (2016):** *Galaxy Bulges and Their Massive Black Holes: A Review*  
  [arXiv:1501.02937v3](https://arxiv.org/abs/1501.02937v3)
- **Marsden, Shankar, Ginolfi & Zubovas (2020):** *The case for the fundamental MBH–σ relation*  
  [arXiv:2004.00098v1](https://arxiv.org/abs/2004.00098v1)


# WHAT: Was beschreibt die M–σ-Beziehung?

- **$M_{\mathrm{BH}}$:** Masse des zentralen supermassereichen Schwarzen Lochs.
- **$\sigma$:** Streuung der Sterngeschwindigkeiten entlang unserer Sichtlinie im Bulge beziehungsweise Sphäroid. Messbereich und Gewichtung müssen definiert sein.
- **Beobachtung:** Größere Geschwindigkeitsdispersion geht tendenziell mit größerer Schwarze-Loch-Masse einher.

$$M_{\mathrm{BH}}\propto\sigma^\beta$$



**So liest man die Grafik:** Beide Achsen sind logarithmisch skaliert. Rot kennzeichnet Core-Sérsic-Systeme, Blau Sérsic-Systeme; Kreuzmarkierungen kennzeichnen Balkengalaxien. Die Linien zeigen getrennte Anpassungen für die Gruppen.

**Bedeutung:** Ein kompaktes Objekt im Zentrum korreliert mit Sternbewegungen auf viel größeren räumlichen Skalen. Das deutet auf verbundene Entwicklung hin. Die Beziehung besitzt Streuung und keine für alle Stichproben festgelegte Steigung.

*Quelle: Graham (2016), Abschnitt 5 und Abbildung 4. Grafik ursprünglich aus Graham & Scott (2013).*


# WHO: Wer hat die Beziehung untersucht?

| Arbeit | Beitrag zur M–σ-Beziehung |
|---|---|
| **Ferrarese & Merritt (2000)** und **Gebhardt et al. (2000)** | Grundlegende Beobachtungsarbeiten. Unterschiedliche anfängliche Steigungen: ungefähr 4,8 beziehungsweise 3,75. |
| **Graham (2016)** | Kritischer Literaturüberblick: erklärt den Einfluss von Regression, Messunsicherheiten und Galaxienstruktur auf Steigung und Streuung. |
| **Marsden et al. (2020)** | Literaturüberblick plus neue Residuen- und Hauptkomponentenanalysen veröffentlichter Messungen, insbesondere aus de Nicola et al. (2019). |

**Was verbindet die beiden Papers?** Beide behandeln M–σ als deutlichen empirischen Zusammenhang zwischen Schwarzem Loch und Wirtsgalaxie.

**Was betonen sie unterschiedlich?** Graham erklärt, warum Stichproben und Methoden unterschiedliche Steigungen liefern. Marsden et al. untersuchen statistisch die Bedeutung von σ und diskutieren Feedback als mögliche physikalische Erklärung.

**Einordnung:** Marsden et al. sammeln für diese Untersuchung keine neuen Teleskopdaten. Eine enge Korrelation allein beweist noch keinen bestimmten Entstehungsmechanismus.

*Quellen: Graham (2016), Abschnitte 5.1–5.2; Marsden et al. (2020), Abschnitte 2 und 4. Die historischen Arbeiten werden anhand dieser Papers zusammengefasst.*


# HOW: Wie wird gemessen und ausgewertet?

**1. Geschwindigkeitsdispersion messen:** Unterschiedliche Dopplerverschiebungen der Sterne verbreitern die Absorptionslinien im gemeinsamen Spektrum. Instrumentelle Verbreiterung und stellare Linienprofile müssen berücksichtigt werden. $\sigma_e$ bezeichnet hier die Dispersion innerhalb des effektiven Radius, der die Hälfte des Lichts einschließt.

**2. Schwarze-Loch-Masse bestimmen:** Dynamische Modelle der Stern- oder Gasbewegungen nahe dem Zentrum liefern die zentrale Masse. Das ist eine andere Messaufgabe als die Dispersion über einen größeren Bereich.

**3. Zusammenhang modellieren:** Die logarithmierten Werte werden mit einem Modell beschrieben. Marsden et al. verwenden unter anderem die **PCA** (Hauptkomponentenanalyse), die die Richtung großer Varianz in den standardisierten Daten bestimmt. Anschließend prüfen sie die Abweichungen vom Modell.



**Oben:** Blaue Messpunkte, rote PCA-Modellgerade, grüner Streubereich.  
**Unten:** Residuen, also beobachtete logarithmische Masse minus logarithmische Modellmasse. Grau ist das Histogramm, rot gestrichelt eine Gaußanpassung.

**Ergebnis dieses Papers:** PCA-Steigung $\beta\approx5{,}4$; Gaußbreite der Massenresiduen etwa **0,47 dex**. Das sind Ergebnisse dieses Modells und dieser Stichprobe, keine universellen Konstanten. Die Residuenbreite ist nicht automatisch die um Messfehler bereinigte intrinsische Streuung.

**Schluss:** M–σ liefert einen Hinweis auf verbundene Entwicklung. Messfehler, Auswahl und Galaxienstruktur beeinflussen die Interpretation. Feedback kann den Zusammenhang erklären, wird durch die Korrelation allein aber nicht bewiesen.

*Quellen: Graham (2016), Abschnitte 3 und 5; Marsden et al. (2020), Abschnitt 2.3, Gleichung 7, Abbildung 5a und Abschnitt 4. Gezeigt ist ausschließlich der M–σ-Teil der Abbildung.*
