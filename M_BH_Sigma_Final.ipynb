# WHAT: Was schauen wir uns an?

- Viele Galaxien besitzen ein **supermassereiches Schwarzes Loch (SSL)** im Zentrum.
- $\sigma$ beschreibt die Streuung der Sterngeschwindigkeiten im Bulge entlang unserer Sichtlinie.
- Größeres $\sigma$ geht im Durchschnitt mit größerer BH-Masse einher.

$$\sigma^2=\langle(v-\langle v\rangle)^2\rangle
\qquad\text{und für }\langle v\rangle=0:\quad\sigma^2=\langle v^2\rangle$$



$$\log_{10}\!\left(\frac{M_{\mathrm{BH}}}{M_\odot}\right)=\alpha+\beta\log_{10}\!\left(\frac{\sigma}{200\,\mathrm{km\,s^{-1}}}\right)$$

Farben: Methoden der Massenbestimmung. Schwarz: einfacher linearer Trend. Die logarithmische Auftragung linearisiert das Potenzgesetz.

*Daten: van den Bosch (2016), ausgewählte dynamische Massenbestimmungen. Literatur: Graham (2016), Abschnitt 5.*

# HOW: Wie werden die Daten gewonnen und analysiert?

- **Dispersion $\sigma$:** Absorptionslinien im gemeinsamen Sternlicht werden durch unterschiedliche Dopplerverschiebungen verbreitert. Ein verbreitertes Vergleichsspektrum wird angepasst; der Instrumenteinfluss wird berücksichtigt.
- **Spektralmodell:** Vergleichsspektrum mit einer Geschwindigkeitsverteilung falten; ihre Breite liefert $\sigma$. Eine einfache Näherung ist eine Gaußverteilung:

$$f(v)=\frac{1}{\sqrt{2\pi}\,\sigma}\exp\!\left[-\frac{(v-v_{\mathrm{sys}})^2}{2\sigma^2}\right]\qquad G=T\ast f+\text{Rauschen}$$

*Spektralmodell schematisch: $T$ ist das Vergleichsspektrum, $f$ die Geschwindigkeitsverteilung und $G$ das beobachtete Spektrum; $\ast$ bezeichnet die Faltung.*

- **BH-Masse:** Zentrale Stern- oder Gasbewegungen räumlich aufgelöst beobachten. Dynamische Modelle mit unterschiedlichen BH-Massen und der übrigen Galaxienmasse mit den Daten vergleichen.
- **Analyse bei Marsden et al.:** Veröffentlichte Messwerte; PCA der standardisierten logarithmierten Daten, anschließend Residuen untersuchen.
